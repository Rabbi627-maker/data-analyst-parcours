1- la consigne demande de remplacer les valeurs manquantes par la médiane , supprimer la colonne cabin, supprimer les lignes ou embarkated est manquant et enrégistrer dans un dattaframe titanic_propre
Qui après cette oppération seras sans valeurs manquantes


```python
titanic_propre = df.fillna({'Age': df['Age'].median()})
titanic_propre = titanic_propre.drop('Cabin', axis=1)
titanic_propre = titanic_propre.dropna(subset=['Embarked'])
titanic_propre.isna().sum()
```
